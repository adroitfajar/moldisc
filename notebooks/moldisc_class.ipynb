# MolDisc classification tutorial

This notebook runs the complete **SMILES-X binary classifier -> GPT generation -> probability screening -> reporting** workflow on the bundled small-data example: 100 labelled and 1,000 unlabelled molecules in `data/data_2`. Start Jupyter Lab from the repository root and run the notebook unchanged once. Then keep the bundled example intact, copy its data folder, and edit the **User settings** block for your own dataset. The deliberately small model and generation settings are an installation check, not a scientific benchmark; see `data/README.md` for dataset provenance and terms.

````python
from pathlib import Path
import json
import sys
import pandas as pd
from IPython.display import Image, display

repo_root = Path.cwd().resolve()
while repo_root.parent != repo_root and not (repo_root / 'moldisc.py').exists():
    repo_root = repo_root.parent
if not (repo_root / 'moldisc.py').exists():
    raise FileNotFoundError('Could not locate the MolDisc repository root.')
sys.path.insert(0, str(repo_root))
from moldisc import moldisc
from rdkit.Chem.Draw import IPythonConsole
# Keep RDKit grid images as saveable PIL objects during the pipeline.
IPythonConsole.UninstallIPythonRenderer()

# USER SETTINGS: keep data/data_2 unchanged; copy it to a new folder, replace both CSVs,
# then change these values. Use -1 to disable a numeric stopping rule.
PROJECT_NAME = 'data_2'             # folder under data/
PROJECT_FOLDER = 'results/tutorial/classification'
DATA_LABEL = 'Positive class'
DATA_UNITS = 'probability'
STOP_AFTER_CYCLES = -1              # disabled in this tutorial
STOP_AFTER_GENERATED = 8            # valid molecules absent from active inputs and prior cycles
TARGET_PROPERTY_VALUE = None        # e.g. 0.9; direction is 'max' below

print('MolDisc repository detected.')
print(f'Python: {sys.version.split()[0]}')
````

````python
labeled = pd.read_csv(repo_root / 'data' / PROJECT_NAME / 'labeled.csv')
unlabeled = pd.read_csv(repo_root / 'data' / PROJECT_NAME / 'unlabeled.csv')
if list(labeled.columns) != ['smiles', 'property']:
    raise ValueError("labeled.csv must have exactly: smiles, property")
if list(unlabeled.columns) != ['smiles']:
    raise ValueError("unlabeled.csv must have exactly: smiles")
if not set(labeled['property']).issubset({0, 1}):
    raise ValueError("classification property values must be 0 or 1")
display(labeled.head())
print(f'{len(labeled)} labeled and {len(unlabeled)} unlabeled molecules')
display(labeled['property'].value_counts().sort_index().rename('count').to_frame())
````

## Run the smoke-scale workflow

This installation test stops after eight valid molecules absent from the active inputs and prior cycles, and uses the neutral `classification_threshold=0.5`. MolDisc writes generated artifacts under `PROJECT_FOLDER` and does not modify the source CSV files. The deliberately tiny model may return no property-selected molecules; that is a valid bounded result. Saved SMILES-X models are reused only when the signed input, cleaned-table, hyperparameter, source-code, and artifact digests all match; unsigned or mismatched predictors are retrained. For real screening, choose one primary endpoint before running (`STOP_AFTER_CYCLES`, `STOP_AFTER_GENERATED`, or `TARGET_PROPERTY_VALUE`), prespecify the allowed-element domain and classification threshold, and never tune the threshold on generated candidates merely to force a non-empty output.

````python
discovered = moldisc(
    project_name=PROJECT_NAME, project_folder=PROJECT_FOLDER,
    data_name='classification_demo', data_label=DATA_LABEL,
    data_units=DATA_UNITS, model_type='classification', scale_output=False,
    augmentation=False, reuse_smilesx_models=True,
    bs_ref=16, lr_ref=3.0, embed_ref=8,
    lstm_ref=8, tdense_ref=8, dense_depth=1,
    k_fold_number=2, n_runs=1, n_epochs=2, n_gpus=0,
    log_verbose=False, train_verbose=0,
    gpt_pretrained_model='distilgpt2', gpt_augmentation=0,
    gpt_initial_epochs=1, gpt_cycle_epochs=1,
    gpt_tr_batch_size=16, gpt_eval_batch_size=16, gpt_patience=2,
    gpt_num_generation=8, gpt_num_attempts=50,
    gpt_generation_batch_size=32, smilesx_inference_augmentation=False,
    gpt_min_carbon_atoms=2, gpt_min_heavy_atoms=3,
    gpt_max_heavy_atoms=50, gpt_max_molecular_weight=700, gpt_max_logp=6,
    gpt_allowed_elements=['B', 'C', 'N', 'O', 'F', 'Si', 'P', 'S', 'Cl', 'Br', 'I'],
    max_generation=-1, max_generated_molecules=STOP_AFTER_GENERATED,
    cycles=STOP_AFTER_CYCLES, target_property_value=TARGET_PROPERTY_VALUE,
    target_property_mode='max', classification_threshold=0.5, sa_score=10,
    option_show=False, num_mol_top=4, num_mol_bottom=4, random_seed=42,
)
display(discovered)
````

````python
run_dir = repo_root / PROJECT_FOLDER / PROJECT_NAME
display_run_dir = run_dir.relative_to(repo_root) if run_dir.is_relative_to(repo_root) else run_dir
print(f'Results directory: {display_run_dir}')
saved = pd.read_csv(run_dir / 'final' / 'generated_smiles.csv')
display(saved)
if (run_dir / 'cycle_metrics.csv').exists():
    metrics = pd.read_csv(run_dir / 'cycle_metrics.csv')
    summary_columns = ['cycle', 'requested', 'generated', 'cumulative_generated', 'selected', 'property_mean', 'property_max', 'generation_validity_rate', 'generation_novelty_rate', 'generation_yield_rate']
    display(metrics[[column for column in summary_columns if column in metrics.columns]])
figure_paths = sorted((run_dir / 'final').glob('top_*_molecules.png'))[:1]
figure_paths += [run_dir / 'final' / 'tanimoto.jpg']
for image_path in figure_paths:
    if image_path.exists():
        display(Image(filename=str(image_path)))
````
